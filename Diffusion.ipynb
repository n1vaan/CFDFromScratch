### Diffusion Equation and Discretizing 2nd Derivatives

The Diffusion Equation is given by: 

$$\frac{\partial u}{\partial t} = \nu \frac{\partial^2 u}{\partial^2 t}$$

To Discretize, consider $\frac{\partial u}{\partial t}$ for some general $u$. 

$$\frac{\partial u}{\partial t} \approx \frac{u(t+\Delta t) - u(t)}{\Delta t}$$

The second partial can be discretized with respect to $\frac{\partial u}{\partial t}$ as

$$\frac{\partial^2 u}{\partial t^2} \approx \frac{\frac{\partial u}{\partial t}(t+\Delta t) - \frac{\partial u}{\partial t}(t)}{\Delta t}$$

Expressing $\frac{\partial^2 u}{\partial t^2}$ with respect to $u$: 

$$\frac{\partial^2 u}{\partial t^2} \approx \frac{\frac{u(t+2\Delta t) - u(t+\Delta t)}{\Delta t} - \frac{u(t+ \Delta t) - u(t)}{\Delta t}}{\Delta t}$$

By simplifying, we obtain
$$\frac{\partial^2 u}{\partial t^2} \approx \frac{u(t+2\Delta t) - 2u(t+\Delta t) + u(t)}{(\Delta t)^2}$$

Therefore, the Diffusion Equation can be discretized as: 

$$\frac{u_i^{n+1} - u_i^n}{\Delta t} = \nu \frac{u_{i+1}^n - 2u_{i}^n + u_{i-1}^n}{(\Delta x)^2}$$

Now, we solve for $u_i^{n+1}$: 
$$u_i^{n+1} = u_i^n + \frac{\nu \Delta t}{(\Delta x)^2}(u_{i+1}^n - 2u_i^n +u_{i+1}^n)


```python
import numpy as np

nx = 41
dx = 2 / (nx - 1)
nt = 20   
nu = 0.3  
sigma = .2 
dt = sigma * dx**2 / nu 

u = np.ones(nx)    
u[int(.5 / dx):int(1 / dx + 1)] = 2  

x_space = np.linspace(0, 2, nx)
t_space = np.arange(nt + 1) * dt

c = (nu * dt) / dx**2


u_prev = u.copy()
evolution = {t_space[0]: u.copy()}
for t in range(1, nt+1):
    for x in range(1,nx-1):
        u[x] = u_prev[x] + c*(u_prev[x+1]-2*u_prev[x]+u_prev[x-1])
    evolution[t_space[t]] = u.copy()
    u_prev = u.copy()



```
